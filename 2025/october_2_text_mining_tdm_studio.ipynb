### Functions



Functions are where the action starts to happen. Functions often expect something to be passed to them, one or more parameters. They then take these parameters and do something with them. I'm vague here because that's the point. These are expressive features!

Functions are also a place where we can talk about how white-space (tabs and spaces) matters in Python. Everything inside the function has to be indented with tabs or spaces. The first line that's back to the normal margin exists outside of function.

We've already used the `print()` function several times. It is built into Python. Similarly, the `len()` function can tell us the length of a name. But we can also create our own.

```python
print(f"`my_name` is {len(my_name)} letters long")

def how_long_is_a_word(word):
  ## Note the two space indent!
  print(f'Your word, "{word}," is {len(word)} letters long')

# Now we call the function and pass in an argument.
how_long_is_a_word("a word")
```

Conditional statements (`if` statements) and loops also use the same kind of indented syntax. We'll get to loops later, but let's incorporate a lot of what we have learned into a tip calculator.

```python
def tip_calculator(total, good_service = True):
  # We are passing in two parameters, "total" and "good_service".
  # We also create a default value for "good_service"
  if good_service:
    total_with_tip = 1.2 * total
  else:
    total_with_tip = 1.18 * total

  # Blank lines don't matter.

  return total_with_tip # this is what the function outputs

# Now we're outside the function and we can call it:

print(f"You owe ${tip_calculator(20, good_service = False)}")
print(f"You owe ${tip_calculator(20)}")
print(f"You owe ${tip_calculator(20, True)}")

# We can use variable and assign return values to variables!
my_total = 40
my_service = 20 > 10
my_total_with_tip = tip_calculator(my_total, good_service = my_service)
print(f"I owe ${my_total_with_tip}")
```

### Lists



So far, we've been working with single things, like a single number or a single string. However, we often get want to work with many things at once. One way of arranging information is with a list. A list is what it sounds like: a list of things, separated by commas, inside brackets (`[]`).

Lists can have different types of things inside (numbers, strings, other lists, etc.) or all the same kind of thing. Lists are iterable, so you can do the same transformation on every element in the list.

In a list, order matters, and you can access each member of the list with its index value. The tricky thing about lists, though, is that the first member of the list has an index value of `0`, not `1`. As such, I tend to talk about the "zeroth" member and then the "oneth" and "twoth" members, to underscore the difference.

```python
# Lists are lists of things surrounded by brackets

my_list = [my_name, 2, "tree", 82.4]
my_list_of_numbers = [3, 5, 2, 8, 29, 3.2]


print(my_list[0]) # get the zeroth member of the list.
print(my_list_of_numbers[-1]) # get the last member of the list.
```

```python
# Lists have lengths"

print(len(my_list))
print(len(my_list_of_numbers))
```

```python
# You can iterate over a list using a `for` loop")
for element in my_list_of_numbers: # Note the colon! We've got to indent
  # "element" is an arbitrary variable that refers to the current list
  # member in the loop
  print(element * 2.1)
```

```python
# You can also iterate over lists using the very Pythonic
# "list comprehension" syntax. This is hard to understand
# at first, but it allows for very pithy yet expressive code

my_list_of_squared_numbers = [n * n for n in my_list_of_numbers]
# above, "n" is doing the same work as "element" in the previous cell

print(my_list_of_squared_numbers)
```

**XML** is a rich, text-based data format that nests information inside tags. It’s somewhat comprable to a deeply structured JSON file or a deeply nested Python dictionary.

In order to parse XML, we can make use of the [LXML](https://lxml.de/) library. Once the XML structure is decoded, we can use the handy [Beautiful Soup](https://pypi.org/project/beautifulsoup4/) library to extract the information, though here we are simply using Beautiful Soup to strip HTML tags.

```python
from lxml import etree
from bs4 import BeautifulSoup

def getxmlcontent(file, strip_html=True):
    try:
        # We are interested in the following fields, so we make a blank
        # dictionary to hold them.
        result = {
            "goid": None,
            "title": None,
            "date": None,
            "publisher": None,
            "pub_title": None,
            "author": None,
            "text": None
        }

        # Parse the file and find the "root" tag, which is at the base of
        # any XML document.
        tree = etree.parse(file)
        root = tree.getroot()

        # Try to find different tags and write the responses to our "result"
        # dictionary.
        if root.find(".//ISOExpansion").text == "English": # Only use English articles
            if root.find('.//GOID') is not None:
                result["goid"] = root.find('.//GOID').text
            if root.find('.//Title') is not None:
                result["title"] = root.find('.//Title').text
            if root.find('.//PubFrosting/Title') is not None:
                result["pub_title"] = root.find('.//PubFrosting/Title').text
            if root.find('.//NumericDate') is not None:
                result["date"] = root.find('.//NumericDate').text
            if root.find('.//PublisherName') is not None:
                result["publisher"] = root.find('.//PublisherName').text
            if root.find('.//Author/NormalizedDisplayForm') is not None:
                result["author"] = root.find('.//Author/NormalizedDisplayForm').text
            # Check for text in various potential places in the XML tree
            if root.find('.//FullText') is not None:
                result["text"] = root.find('.//FullText').text
            elif root.find('.//HiddenText') is not None:
                result["text"] = root.find('.//HiddenText').text
            elif root.find('.//Text') is not None:
                result["text"] = root.find('.//Text').text

            # Strip html from text portion
            if result["text"] is not None and strip_html == True:
                result["text"] = (BeautifulSoup(result["text"])
                  .get_text()
                  .replace('\n', ' ')
                  .replace('\\', '')
                  .strip())

    except Exception as e:
        print(f"Error while parsing file {file}: {e}")

    return result
```

#### Collect Information about the Dataset

With the proof of concept in the bag, we can write a function that iterates over every file in the folder, parses and extracts the content, and dumps everything into a DataFrame. I am using [Polars](https://pola.rs) here, but you can also use [pandas](https://pandas.pydata.org/). Both are great and rely on somewhat different conceptual models, but they both get you where you need to go, with Polars getting the edge for particularly large datasets.

```python
import os
import random
from tqdm.notebook import tqdm

def parse_files(folder, sample_size = None):
    # Get the list of files.
    files = os.listdir(folder)

    # If we set a sample size, pick a sample of file names.
    if sample_size:
        files = random.sample(files, sample_size)

    # Create a list with the results of each file.
    rows = []

    # Iterate over the files
    for file in tqdm(files):
        result = getxmlcontent(f"{folder}/{file}", strip_html=True)
        rows.append(result)

    return rows
```

```python
data = parse_files("PLosAIColumbia")

print(f"The length of the data is {len(data)}")
print(f"The data have the following keys:\n{data[0].keys()}")
```

And now we can feed data into Polars.

```python
import polars as pl

df = pl.DataFrame(data)

df.head()
```

```python
# Convert dates to date objects
df = df.with_columns(
    pl.col("date").cast(pl.String).str.strptime(pl.Date, "%Y-%m-%d"),
    pl.col("text").str.split(" ").list.len().alias("word_count")
)
```

```python
df["word_count"].describe()
```

```python
import altair as alt

chart = (
  alt.Chart(df)
    .mark_bar()  # Use mark_bar for histograms
    .encode(
        alt.X("year(date):T", title="Date"),
        alt.Y("count()", title="Frequency")
    )
    .properties(
        title="Distribution of Article Publication Dates"
    )
 )

chart.show()
```
